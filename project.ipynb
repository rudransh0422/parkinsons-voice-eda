```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


url = "https://archive.ics.uci.edu/ml/machine-learning-databases/parkinsons/parkinsons.data"
df = pd.read_csv(url)

# Optional: save a local CSV copy inside your Colab session
```

```python
df.shape
df.head()
df.tail()
```

```python
#question 1
df['status'].value_counts()
```

```python
#Q2 Calculate mean,median,minimum and maximum for selected voice measurements such as MDVP:FO(HZ),MDVP:JITTER(%) and MDVP:SHIMMER
fo = np.array(df["MDVP:Fo(Hz)"])
print("MDVP:Fp(Hz)")
print("Mean:", np.mean(fo))
print("Median:", np.median(fo))
print("Min:", np.min(fo))
print("Max:", np.max(fo))

jitter = np.array(df["MDVP:Jitter(%)"])
print("MDVP:Jitter(%)")
print("Mean:", np.mean(jitter))
print("Median:", np.median(jitter))
print("Min:", np.min(jitter))
print("Max:", np.max(jitter))

shimmer = np.array(df["MDVP:Shimmer"])
print("MDVP:Shimmer")
print("Mean:", np.mean(shimmer))
print("Median:", np.median(shimmer))
print("Min:", np.min(shimmer))
print("Max:", np.max(shimmer))
```

```python
#q3Compare average voice measurements grouped by status
selected_cols = ["MDVP:Fo(Hz)", "MDVP:Jitter(%)", "MDVP:Shimmer"]
df.groupby("status")[selected_cols].mean()
```

```python
#Q4 find 10 highest values for one selected measurement
df['MDVP:Fo(Hz)'].nlargest(10)
```

```python
#sorting in descending order
df.sort_values(by="MDVP:Fo(Hz)", ascending=False).head(10)
```
